**Write a program to count the frequency of each character in a string using a dictionary.**

```python
str1 = "Programming"
dict = {}
for i in str1:
    if i in dict:
        dict[i] +=1
    else:
        dict[i] = 1
print(dict)
```

**Write a program to find the key having the maximum value in a dictionary.**

```python
dict1 = {'A':95, 'B':92, 'C':78, 'D':90}
largest_value = 0
for i in dict1:
    if dict1[i] > largest_value:
        largest_value = dict1[i]
        largest_key = i
print("Largest Key : ", largest_key)
```

**Write a program to remove duplicate values from a dictionary.**

```python
dict1 = {'a': 10, 'b': 20, 'c': 10, 'd': 30, 'e': 40} 
result_dict = {}
seen_values = set()

for i, j in dict1.items():
    if j not in seen_values:
        result_dict[i] = j
        seen_values.add(j)

print(result_dict)
```

**Write a program to find the sum of all values in a dictionary.**

```python
marks = {'Math':85, 'Science':90, 'English':80} 
sum = 0
for i,j in marks.items():
    sum += j

print("Total Marks : ", sum)
```

**Write a program to create a dictionary from two lists: one containing keys and the other containing values.**

```python
Keys = ['A', 'B', 'C'] 
Values = [10, 20, 30] 
result = {}
for i in Keys:
    for j in Values:
        result[i]=j
print(result)
```
