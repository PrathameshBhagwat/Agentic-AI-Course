3. Bank Account — Encapsulation
Create a BankAccount class.
• Create private variable __balance.
• Create deposit() and withdraw() methods.
• Create get_balance() getter.
• Don't allow withdrawal greater than balance.
• Don't allow negative deposit.
• Keep balance inaccessible directly from outside. 

```python
class BankAccount:
    def __init__(self):
        self.__balance=0
        
    def menu(self):
        print("1.Check Balance")
        print("2.Deposit")
        print("3.Withdraw")
        print("4.Exit")
        
        while True: 
            choice = int(input("Enter you Choice : "))
            
            match choice:
                case 1:
                    self.get_balance()
            
                case 2:
                    self.deposit()

                case 3:
                    self.withdraw()
                    
                case 4:
                    print("Exit")
                    break
            
    def get_balance(self):
        return self.__balance
    
    
    def deposit(self):
        amount = int(input("Enter the ammount to deposit : "))
        if amount > 0:
            self.__balance += amount
            print("Amount after deposit : ",self.__balance)
        else:
            print("Enter the valid amount.")
        
    def withdraw(self):
        amount = int(input("Enter the ammount to withdraw : "))
        if amount < self.__balance:
            self.__balance -= amount
            print("Amount after the withdrwal : ",self.__balance)
        else:
            print("Insufficient balance.")
            
acc1 = BankAccount()
acc1.menu()
```

• Create a method start().
• Create child classes Car, Bike, and Bus.

10. Calculator — Method Overloading
Create a Calculator class.
Create an add() method that should support:
add(10, 20)
add(10, 20, 30)
add(10, 20, 30, 40)
• Use *args or default arguments.
• Do not create multiple add() methods.
• Similarly implement multiply() using *args. 

```python
class Calculator:
    def add(self,num1=0,num2=0,num3=0,num4=0):
        return num1+num2+num3+num4
    
    def mul(self,*args):
        total = 1 
        for i in args:
            total *= i
        
        return total
    
cal = Calculator()
cal.add(10,20,30,40)
cal.mul(10,20,10)
```

```python
list1 = [10, 20, 30, 40, 50]
list2 = [20, 40, 60, 80, 100]
result = []

for i in list1:
    for j in list2:
        if i == j:
            result.append(i)
            
print(result)
```

Write a Python program to find the first non-repeated character in a string.

```python
str1 = "Prathamesh"
frequency = {}
result = []

for i in str1:  
    frequency[i] = frequency.get(i,0)+1
    
for char,freq in frequency.items():
    if freq == 1:
        result.append(char)
        
print(result[0])
```

Write a Python program to find the second largest unique number from a list without
using sort(). 

```python
list1=[20,30,10,70,80,85,90]

largest = 0 
second_largest = 0

for i in list1:
    if i > largest:
        second_largest = largest
        largest = i  
    elif i<largest and i> second_largest:
        second_largest = i
        

print(second_largest)
```
