```python
class Room:
    def __init__(self,name):
        self.name = name
        
    def display(self):
        print("Room is ,",self.name)
        
class House:
    def __init__(self):
        self.bedroom = Room("Bedroom")
        self.kitchen = Room("Kitchen")
        
    def rooms_in_house(self):
        self.bedroom.display()
        self.kitchen.display()
    
house = House()
house.rooms_in_house()
```
