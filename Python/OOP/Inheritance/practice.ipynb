**Employee System**

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    
    def display(self):
        print("Name : ",self.name)
        print("Salry : ", self.salary)

class Developer(Employee):
    def __init__(self,name,salary,designation,coding):
        super().__init__(name,salary)
        self.designation = designation
        self.coding = coding
    
    def display(self):
        print("Name : ",self.name)
        print("Salry : ", self.salary)
        print("Designation : ", self.designation)
        print("I work on the ",self.coding," project")
        

class Manager(Employee):
    def __init__(self,name,salary,designation,team_size):
        super().__init__(name,salary)
        self.designation = designation
        self.team_size = team_size
    
    def display(self):
        print("Name : ",self.name)
        print("Salry : ", self.salary)
        print("Designation : ", self.designation)
        print("Manage the team  of ",self.team_size," people")  
        
    
obj1 = Manager("Sarthak",40000,"Manager",10)
obj1.display()
```
