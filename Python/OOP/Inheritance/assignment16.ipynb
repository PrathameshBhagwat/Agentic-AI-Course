**Create a Person parent class with name and age. Create a Student child class with**
**roll_no and marks. Use super() to initialize the parent attributes.**

**Create an Employee parent class with name and salary. Create a Manager child class**
**with an additional department attribute. Display all details using constructors.**

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    
    def display(self):
        print("Name : ",self.name)
        print("Salry : ", self.salary)
        
        
class Manager(Employee):
    def __init__(self,name,salary,designation,team_size):
        super().__init__(name,salary)
        self.designation = designation
        self.team_size = team_size
    
    def display(self):
        print("Name : ",self.name)
        print("Salry : ", self.salary)
        print("Designation : ", self.designation)
        print("Manage the team  of ",self.team_size," people")  
        
    
obj1 = Manager("Sarthak",40000,"Manager",10)
obj1.display()
```

**Create a BankAccount parent class with account_number and balance. Create a**
**SavingsAccount child class with an additional interest_rate. Use super() in the child**
**constructor.**

```python
class BankAccount:
    def __init__(self,account_no,balance):
        self.account_no = account_no
        self.balance = balance
    
class SavingsAccount(BankAccount):
    def __init__(self,account_no,balance,interest_rate):
        super().__init__(account_no,balance)
        self.interest_rate = interest_rate
        
    def display(self):
        print("Account no : ",self.account_no)
        print("Balance in the account : ",self.balance)
        print("Interest on the amount is", self.interest_rate)
        

obj1 = SavingsAccount(12345,300000,7.8)
obj1.display()
```

**Create a parent class Company with a company_name attribute. Create two child**
**classes Developer and Tester, each with their own additional attributes and methods.**


```python
class Company:
    def __init__(self,com_name):
        self.com_name=com_name
        
    def work(self):
        print("Working in the :",self.com_name)
        
        
class Developer(Company):
    def __init__(self,com_name,name,profession,project):
        super().__init__(com_name)
        self.name = name
        self.profession = profession
        self.project = project
    
    def work(self):
        super().work()
        print(f"My name is {self.name}, and I work as the {self.profession} on the {self.project}")
    
    
class Tester(Company):
    def __init__(self,com_name,name,profession,project):
        super().__init__(com_name)
        self.name = name
        self.profession = profession
        self.project = project
    
    def work(self):
        super().work()
        print(f"My name is {self.name}, and I work as the {self.profession} on the {self.project}")

developer = Developer("Infystent","Swapnil","Developer","Banking Project")        
tester = Tester("Infystent","Sarthak","Tester","Banking Project")

developer.work()
tester.work()
```

**Bank Account**
**Create a BankAccount class with a private attribute __balance.**
**Create get_balance() to view the balance.**
**Create set_balance() to update the balance only if the amount is valid.**
**Create a withdraw() method with proper balance validation.**

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0
        
    def get_balance(self):
        return self.__balance
    
    def set_balance(self,balance):
        if balance > 0:
            self.__balance += balance   
        print("Current amount in account : ",self.__balance)
        
    def withdraw(self,balance):
        if balance < self.__balance:
            self.__balance -= balance
        else:
            print("Insuffiecient balance to withdraw")
        
        print("Current amount in account : ",self.__balance)

balance = int(input("Enter amount to add"))
user = BankAccount()
user.set_balance(balance)
withdrawal = int(input("withdraw"))
user.withdraw(withdrawal)
```

**Constructor + Private Data**
**Create a User class with private __password.**
**Initialize it through the constructor.**
**Create get_password() and set_password().**
**Setter should update the password only if it contains at least 6 characters.**

```python
class User:
    def __init__(self,password):
        self.set_password(password)
        
    def get_password(self):
        return self.__password
        
    def set_password(self,password):
        if len(password) >= 6:
            self.__password = password
        else:
            print("Enter password more than 6 characters")

password = input("Enter the password : ")
user1 = User(password)
user1.get_password()
```

**Multilevel Inheritance + Getter/Setter**
**Person → private age with getter/setter.**
**Employee → salary.**
**Manager → department.**
**Use super() at each level**

```python
class Person:
    def __init__(self,name,age):
        self.set_name_age(name,age)
    
    def get_name_age(self):
        return f"Name is {self.__name} and age {self.__age}"
    
    def set_name_age(self,name,age):
        self.__name = name
        self.__age = age

class Employee(Person):
    def __init__(self, name, age,salary):
        super().__init__(name, age) 
        self.set_salary(salary)
    
    def get_salary(self):
        return f"Salary : {self.__salary}"
    
    def set_salary(self,salary):
        self.__salary = salary
    

class Manager(Employee):
    def __init__(self, name, age, salary,department):
        super().__init__(name, age, salary)
        self.set_department(department)
        
    def get_department(self):
        return f"Department : {self.__department}"
    
    def set_department(self,department):
        self.__department = department
        
user1 = Manager("Sarthak",22,70000,"IT")
user1.get_name_age()
user1.get_salary()
user1.get_department()
```
