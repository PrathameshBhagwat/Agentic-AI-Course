**Addition of the three objects**

```python
class Demo:
    def __init__(self,num):
        self.num = num
    
    def __add__(self,other):
        return self.num + other.num
    
obj1 = Demo(10)
obj2 = Demo(20)
obj3 = Demo(30)
result = (obj1 + obj2) + obj3.num
print(result)
```

```python
class Demo:
    def __init__(self,num):
        self.num = num
        
    def __add__(self, other):
        return Demo(self.num + other.num)
    
    
    def __str__(self):
        return str(self.num)
    
obj1 = Demo(10)
obj2 = Demo(20)
obj3 = Demo(30)
result = (obj1+obj2) + obj3

print(result)
```
