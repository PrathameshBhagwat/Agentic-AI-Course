**Student Information System Create a Student class with:**
**• name**
**• roll_no**
**• branch**
**• marks** 
**Create 3 student objects and create a method display() to print their details.**

```python
class StudentInfo:
    def __init__(self, name, roll_no, branch, marks):
        self.name=name
        self.roll_no=roll_no
        self.branch=branch
        self.marks=marks
    
    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)
        print("Branch:", self.branch)
        print("Marks:", self.marks)

Student1=StudentInfo("Gagan", 8, "BE", 95)
Student2=StudentInfo("Swapnil", 15, "BE", 90)
Student3=StudentInfo("Sarthak", 17, "BE", 85)
Student4=StudentInfo("Omkar", 10, "BE", 80)
Student1.display()
Student2.display()
Student3.display()
Student4.display()
```

**Employee Salary Create an Employee class with: • name  • employee_id  • salary  Create a method display_salary() that displays the employee's salary. Create 3 employee objects. : Don't assign the values directly inside the class. Pass them while creating objects.**

```python
class EmployeeInfo:
    def __init__(self, name, emp_id, department, salary):
        self.name=name
        self.emp_id=emp_id
        self.department=department
        self.salary=salary
    
    def display(self):
        print("Name:", self.name)
        print("Employee ID:", self.emp_id)
        print("Department:", self.department)
        print("Salary:", self.salary)
        
Employee1=EmployeeInfo("Gagan", 101, "IT", 50000)
Employee2=EmployeeInfo("Swapnil", 102, "IT", 60000)
Employee3=EmployeeInfo("Sarthak", 103, "HR", 70000)
Employee4=EmployeeInfo("Omkar", 104, "IT", 80000)
Employee1.display()
Employee2.display()
Employee3.display()
Employee4.display()
```

**3. Mobile Phone Create a Mobile class with: • brand  • model  • price  Create a method: display_details() Create 3 mobile objects with different information. Expected concept: Constructor + objects.** 

```python
class Mobile:
    def __init__(self, brand, model, price):
        self.brand=brand
        self.model=model
        self.price=price
    
    def display(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Price:", self.price)

Mobile1=Mobile("Realme", "6i", 13000)
Mobile2=Mobile("Samsung", "Galaxy A54S", 30000)
Mobile3=Mobile("Realme", "Narzo 10", 13000)
Mobile4=Mobile("Oppo", "Reno 7", 30000)
Mobile1.display()
Mobile2.display()
Mobile3.display()
Mobile4.display()
```
