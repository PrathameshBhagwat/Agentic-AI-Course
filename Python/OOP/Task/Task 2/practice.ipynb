```python
list1 = [10, 20, 30, 40, 50]
list2 = [20, 40, 60, 80, 100]
result = []

for i in list1:
    for j in list2:
        if i == j:
            result.append(i)
            
print(result)
```

using sort().

10. Given two dictionaries containing product names and prices, merge them into one
dictionary. If a product exists in both, keep the higher price. 

```python
product1 = {"Laptop":50000, "Mobile":20000, "Tablet":30000}
product2 = {"Laptop":40000,"SmartWatch":2000}
result = {}
for i in product1:
    result[i] = product1[i]
    
for i in product2:
    if i in result:
        if product2[i]>result[i]:
            result[i] = product2[i]
            
    else:
        result[i] = product2[i]
        
print(result)
```

11.Create a function calculator(a, b, operation) that performs addition, subtraction,
multiplication, and division based on the given operation. 

```python
def calculator(a,b,operation):
    if operation == "+":
        return a+b
    elif operation == "-":
        return a-b
    elif operation == "*":
        return a*b
    elif operation == "/":
        return a/b
    
num1 = int(input("Enter the number : "))
num2 = int(input("Enter the number : "))
operation = input("Enter the operation : ")

calculator(num1,num2,operation)
```

12.Create a function that accepts a list of numbers and returns the largest number,
smallest number, sum, and average.

```python
def analysis(numbers):
    largest = numbers[0]
    minimum = numbers[0]
    sum = 0 
    
    for i in numbers:
        if i > largest:
            largest = i

        if i < minimum:
            minimum = i
            
        sum = sum + i

    average = sum / len(numbers)
    
    return largest, minimum, sum, average

numbers = [10,20,30,40,60,50]
largest, minimum, sum, average = analysis(numbers)
print("Largest : ",largest)
print("Smallest : ",minimum)
print("Total : ",sum)
print("Average : ",average)
```

13.Create a BankAccount class with account holder, account number, and balance.
Add methods for deposit, withdrawal, and displaying balance. Prevent withdrawal
when the balance is insufficient.

```python
class BankAccount:
    def __init__(self,name,ac_number,balance):
        self.name = name 
        self.ac_number = ac_number
        self.balance = balance
        
    def deposit(self):
        amount = float(input("Enter the amount : "))
        self.balance += amount
    
    def withdrawal(self):
        amount = float(input("Enter the amount : "))  
        if amount > self.balance:
            print("Balance is insufficient.")
        else:
            self.balance -= amount
        
    def display_amount(self):
        print("Amount : ",self.balance)
        
obj1 = BankAccount("Prathamesh",123456,500)
obj1.deposit()
obj1.withdrawal()
obj1.display_amount()
```

14.Create an Employee class with employee name, ID, department, and salary. Add
methods to display details, increase salary by a given percentage, and check
whether the salary is above ₹50,000. 

```python
class Employee:
    def __init__(self,name,ID,department,salary):
        self.name = name
        self.ID = ID
        self.department = department
        self.salary = salary
        
    def display(self):
        print("Name : ",self.name)
        print("ID : ",self.ID)
        print("Department : ",self.department)
        print("Salary : ",self.salary)
        
    def increase_salary(self):
        percentage = float(input("Enter the percentage to increase : ")) 
        increament = self.salary * percentage / 100
        self.salary = self.salary + increament
        print("New Salary : ",self.salary)
        
    def check_salary(self):
        if self.salary>50000:
            print("Salary is above 50,000")
        else:
            print("Salary is less than 50,000")
            
employee1 = Employee("Prathamesh", 123, "IT", 60000)
employee1.display()
employee1.increase_salary()
employee1.check_salary()
```

15.Create a Vehicle class with brand and speed. Create Car and Bike child classes with
additional attributes. Use inheritance and method overriding to display their
complete details. 

```python
class Vehicle:
    def __init__(self,brand,speed):
        self.brand = brand
        self.speed = speed
        
    def display(self):
        print("Brand : ",self.brand)
        print("Speed of the vehicle :", self.speed)
    
    def start(self):
        print("Vehicle has been started")
        
class Car(Vehicle):
    def __init__(self, brand, speed, wheels, price):
        super().__init__(brand, speed)
        self.wheels = wheels
        self.price = price

    def display(self):
        print("Brand : ",self.brand)
        print("Speed of the car :", self.speed)
        print("Wheels in the car : ",self.wheels)  
        print("Price of the car : ",self.price)  

    def start(self):
        print("Car has been started")
        

class Bike(Vehicle):
    def __init__(self, brand, speed, wheels, price):
        super().__init__(brand, speed)
        self.wheels = wheels
        self.price = price
        
    def display(self):
        print("Brand : ",self.brand)
        print("Speed of the bike :", self.speed)
        print("Wheels in the bike : ",self.wheels)  
        print("Price of the bike : ",self.price)  
        
    def start(self):
        print("Bike has been started")
        
v1 = Vehicle("Tata",160)
v1.display()
v1.start()

c1 = Car("Tata Harrier",160, 4,"16L")
c1.display()
c1.start()

b1 = Bike("Kawasaki",240,2,"9L")
b1.display()
b1.start()
```
