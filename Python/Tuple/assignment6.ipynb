**Write a program to find the largest and smallest element in a tuple without using max() and min().**

```python
tuple=(20,27,18,90,70)
sorted_tuple=sorted(tuple)
largest=sorted_tuple[-1]
smallest=sorted_tuple[0]
print("Largest element is:", largest)
print("Smallest element is:", smallest)
```

**Write a program to count the number of even and odd numbers in a tuple**

```python
tuple=(20,27,18,90,70)
even = 0
odd = 0

for i in tuple:
    if i%2==0:
        even+=1
    else:
        odd+=1
print("Number of even numbers:", even)
print("Number of odd numbers:", odd)
```

**Write a program to find all common elements between two tuples.**

```python
tuple1=(20,27,18,90,70)
tuple2=(10,15,25,90,70)
set1=set(tuple1)
set2=set(tuple2)
common_elements=set1.intersection(set2)
common_elements_tuple=tuple(common_elements)
print("Common elements:", common_elements_tuple)
```

**Write a program to create a new tuple containing only positive numbers from a given tuple.**

```python
numbers=(-1,-5,6,7,8,-10,20,-30,40,50)
result = []
for i in numbers:
    if i > 0:
        result.append(i)

result_tuple = tuple(result)
print("Tuple with positive numbers:", result_tuple)
```

**Write a program to remove duplicate elements from a tuple while preserving the original order.**

```python
numbers=(1,2,3,3,5,4,3,2,1,8,0)
result = []

for i in numbers:
    if i not in result:
        result.append(i)
        
result_tuple = tuple(result)
print(result_tuple)
```

**Write a program to separate all even and odd elements of a tuple into two different tuples**

```python
numbers=(10,15,20,25,30)
even = []
odd = []

for i in numbers:
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)
even_tuple = tuple(even)
odd_tuple = tuple(odd)
print("Even numbers tuple:", even_tuple)
print("Odd numbers tuple:", odd_tuple)
```

**Write a program to find the intersection of two tuples**

```python
tuple1=(20,27,18,90,70)
tuple2=(10,15,25,90,70)
set1=set(tuple1)
set2=set(tuple2)
common_elements=set1.intersection(set2)
common_elements_tuple = tuple(common_elements)
print("Intersection : ", common_elements_tuple)
```

**Write a program to find the union of two tuples (unique elements only)**

```python
tuple1=(20,27,18,90,70)
tuple2=(10,15,25,90,70)
set1=set(tuple1)
set2=set(tuple2)
union_elements=set1.union(set2)
union_elements_tuple = tuple(union_elements)
print("Union : ", union_elements_tuple)
```

**Write a program to convert a nested tuple like:**

```python
numbers=((1,2), (3,4), (5,6))
result = []

for i in numbers:
    for j in i:
        result.append(j)

result_tuple = tuple(result)
print("Output : ", result_tuple)
```
