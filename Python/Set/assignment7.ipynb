**Write a program to find the union of two sets without using the union() method**

```python
set1={1,2,3,4,5}
set2={4,5,6,7,8}

union_set= set()
for i in set1:
    union_set.add(i)
for j in set2:
    union_set.add(j)
print("Union of two sets is:", union_set)
```

**Write a program to remove all duplicate elements from a list using a set.**

```python
list1=[1,2,3,4,5,6,5,4,3,2,1]
set1=set(list1)
print("Duplicates removed from list:", set1)
```

**Write a program to find all common elements between two lists using sets.**

```python
list1=[1,2,3,4,5]
list2=[4,5,6,7,8]
set1=set(list1)
set2=set(list2)
common_elements=set1.intersection(set2)
print("Common elements:", common_elements)
```

**Write a program to count the number of unique elements in a list using a set.**

```python
list1=[1,2,3,4,5,4,3,2,1]
set1=set(list1)
count= 0
for i in set1:
    count += 1
print("Number of unique elements in the list:", count)
```

**Write a program to find the missing numbers between 1 to N using sets**

```python
input_numbers = int(input("Enter a number: "))
input_set = set()
for i in range(1, input_numbers + 1):
    input_set.add(i)

numbers =  {1,2,3,5,7,8,10}
missing_numbers = input_set.difference(numbers)
print("Missing numbers:", missing_numbers)
```

**Write a program to create a new set containing only the even numbers from a given set.**

```python
set1={1,2,3,4,5}
even_numbers = set()
for i in set1:
    if i % 2 == 0:
        even_numbers.add(i)
print("Even numbers:", even_numbers)
```

**Write a program to merge three sets into a single set containing only unique elements.**

```python
set1={1,2,3,4,5}
set2={4,5,6,7,8}
set3={7,8,9,10}
result_set = set1.union(set2).union(set3)
print("Result set:", result_set)
```

**Given two strings, find the common characters using sets.**

```python
str1 = "Hello"
str2 = "World"
set1 = set(str1)
set2 = set(str2)
intersection_set = set1.intersection(set2)
print("Intersection of characters in two strings:", intersection_set)
```

**Given multiple lists, find all elements that appear in every list using sets.**

```python
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 2, 3]
list3 = [7, 2, 9, 3]
set1 = set(list1)
set2 = set(list2)
set3 = set(list3)
common_elements = set1.intersection(set2).intersection(set3)
print("Common elements in three lists:", common_elements)
```
