**Find the median of the list**

```python
list1 = [10,11,12]
list2 = [13,14,15]

list1.extend(list2)

if   len(list1) % 2 == 1:
        middle = len(list1) // 2
        print("Middle element:", list1[middle])
        
else: 
    middle = len(list1) // 2
    previous = middle - 1
    median = (list1[previous]+list1[middle])/2
    print("Middle elements:", median)
    
```

**Move Zeroes to last of the list**

```python
list1 = [0,1,0,3,12]
result = []
count = 0
for i in list1:
    if i != 0:
        result.append(i)
    else:
        count += 1
        
for i in range(count):
    result.append(0)
    
print(result)
```

**Add the last substring to the front of the string**

```python
str1 = "aaacbbccc"
list1 = str1.split()
# print(list1)
remaining = []
sub_list =[]

last_char = str1[-1]

for i in range(len(str1)-1,-1,-1):
    if str1[i] == last_char:
        sub_list.append(str1[i])
    else:
        break

start = len(str1) - len(sub_list)
    
for i in range(start):
    remaining.append(str1[i])
    
result_list = sub_list+remaining
print("".join(result_list))
```

**Reverse the set of the 4 words in the string**

```python
str1 = "abcdefghijkl"
result = []

if len(str1) % 4 ==0:
    for i in range(0,len(str1),4):
        word=(str1[i:i+4])
        reverse_word=word[::-1]
        result.append(reverse_word)
          
else:
    print("Characters in the string are not the mulriples of the 4.")
    
result_str = "".join(result)
print(result_str)
```
