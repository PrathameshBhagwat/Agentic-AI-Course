**Find the median of the list**

```python
list1 = [10,11,12]
list2 = [13,14,15]

list1.extend(list2)

if   len(list1) % 2 == 1:
        middle = len(list1) // 2
        print("Middle element:", list1[middle])
        
else: 
    middle = len(list1) // 2
    previous = middle - 1
    median = (list1[previous]+list1[middle])/2
    print("Middle elements:", median)
    
```

**Check the word in the string**

```python
str1 = "Is the good person,and he is talented."
str2 = "he"

if str2 in str1:
    for i in range(len(str1)):
        if str1[i:i+len(str2)] == str2:
            print("The word found at index", i)
            break
        
else:
    print("The word is not found in the string.")
```
