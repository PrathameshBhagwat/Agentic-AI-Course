**Print numbers from 1 to 100 using a for loop.**

```python
for i in range(1 , 101):
    print(i)
```

**Print numbers from 100 to 1 using a while loop.**

```python
i=100
while i > 0:
    print(i)
    i -= 1
```

**Print all even numbers between 1 and 100.**

```python
for i in range(1 , 101):
    if i % 2 ==0:
        print(i)
```

**Print all odd numbers between 1 and 100.**

```python
for i in range(1 , 101):
    if i % 2 ==1:
        print(i)
```

**Print the multiplication table of any number entered by the user.**

```python
num = int(input("Enter a number: "))
for i in range(1,11):
    print(num, "X", i, "=", num*i)
```

**Nested Loops**

**Print a 5 × 5 matrix of stars.**

```python
row = 5
for i in range(row):
    for j in range(row):
        print("* ", end="")
    print()
```

**Print the following pattern.**

```python
row = 5
for i in range(row):
    for j in range(row):
        print(j+1, " ", end="")
    print()
```

**Print the following pattern.**

```python
row = 5
for i in range(row,0,-1):
    for j in range(i):
        print("* ", end="")
    print()
```

```python
row = int(input("Enter the number of rows: "))

for i in range(row):
    for j in range(i+1):
        print(i+1, end="")
    print()
```

**Print the following Floyd's Triangle.**

```python
row = int(input("Enter the number of rows: "))

num=1

for i in range(row):
    for j in range(i+1):
        print(num," ",end="")
        num += 1
    print()
```

```python
n = 20
spaces = int((n-2-1)/2)
star = 1
height =  int(n/2)
hollow_spaces = 0

print("*" * n)


for i in range(height -2):
    print("*",end="")
    print(" " * spaces,end="")
    if i==0:
        print("*",end="")
    else:
        print("*", end="")
        print(" "*hollow_spaces,end="")
        print("*", end="")
    
    print(" "*(spaces+1),end="")
    print("*")
    
    spaces -= 1
    if i==0:
        hollow_spaces = 1
    else:
        hollow_spaces += 2
    
print("*" * n)
```
