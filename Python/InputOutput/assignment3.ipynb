**Write a program to take a student's name, age, city, and percentage as input and display them.**

```python
student_name = input("Enter your name: ")
age = input("Enter your age: ")
city = input("Enter your city: ")
percentage = input("Enter your percentage: ")

print("Student Name:", student_name)
print("Age:", age)
print("City:", city)
print("Percentage:", percentage)
```

**Take the length and width of a rectangle as input and calculate: Area and Perimeter**

```python
length = float(input("Enter the length of the rectangle: "))
breadth = float(input("Enter the breadth of the rectangle: "))
perimeter = 2 * (length + breadth)
area = length * breadth
print("Perimeter of the rectangle:", perimeter)
print("Area of the rectangle:", area)
```

**Take a student's marks  Python, Java, and SQL as input and calculate: Total,Average,Percentage**

```python
marks_python = float(input("Enter marks in Python: "))
marks_java = float(input("Enter marks in Java: "))
marks_sql = float(input("Enter marks in SQL: "))

total_marks = marks_python + marks_java + marks_sql
average_marks = total_marks / 3
percentage = (total_marks / 300) * 100
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
print("Percentage:", percentage)
```
