**Write a program to handle ZeroDivisionError when dividing two numbers entered by the user.** 

```python
try:
    num1 = int(input("Enter the first number: "))
    num2 = int(input("Enter the second number: "))
    result = num1 / num2
    print("Result is ", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```

**Write a program to handle ValueError when the user enters a non-numeric value for age.**

```python
try:
    age = int(input("Enter your age: "))
    print("Age is ", age)
except ValueError:
    print("Error: Invalid input. Please enter a valid integer for age.")
    
```

**Write a program to handle IndexError while accessing an element from a list.** 

```python
list1 = [1, 2, 3, 4, 5]
try:
    index = int(input("Enter the index which has to be accessed: "))
    print(list1[index])
except IndexError:
    print("Error: Index out of range. Please enter a valid index.")
```

**Write a program to handle KeyError when accessing a key from a dictionary.**

```python
dictionary = {"name": "Prathamesh", "age": 22, "city": "Pune"}
key = input("Enter the key to access the dictionary: ")
try:
    print(dictionary[key])
except KeyError:
    print("Error: Key not found in the dictionary. Please enter a valid key.")
```

**Write a program to handle TypeError by attempting to add an integer and a string.**

```python
num1 = int(input("Enter the first number: "))
num2 = input("Enter the second number: ")
try:
    num3 = num1 + num2
    print("Result is ", num3)
except TypeError:
    print("Error: Invalid input type. Please enter valid numbers.")
```

**Write a program that demonstrates the use of try, except, else, and finally blocks.**

```python
try:
    num1 = int(input("Enter the first number: "))
    num2 = int(input("Enter the second number: "))
    result = num1 / num2 
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
else:
    print("Result is ", result)
finally:
    print("Execution completed.")
    
```

**Write a program that catches any exception using except Exception as err and displays:**

```python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

    print("Result:", result)

except Exception as err:
    print("An error occurred:", err)
```
