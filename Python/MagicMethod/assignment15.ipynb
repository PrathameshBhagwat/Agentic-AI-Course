**Product Class – __eq__()**
**Create a Product class with name and price. Implement __eq__() to check whether two**
**products have the same price.**

```python
class Product:
    def __init__(self,name,price):
        self.name = name 
        self.price = price 
        
    def __eq__(self,other):
        return self.price == other.price
    
obj1 = Product("Oil",2000)
obj2 = Product("Oil2", 2000)
obj1 == obj2
        
```

**Shopping Cart – __len__()**
**Create a ShoppingCart class containing a list of products. Implement __len__() so that**

```python
class Cart:
    def __init__(self):
        self.list1 = ["RAM","ROM","Monitor"]
        
    def __len__(self):
        return len(self.list1)
    
cart = Cart()
len(cart)
```

**Bank Account – __str__()**
**Create a BankAccount class with account holder name, account number, and balance.**
**Implement __str__() to display account information when the object is printed.**

```python
class BankAccount:
    def __init__(self,name,account_no,balance):
        self.name = name
        self.account_no = account_no
        self.balance = balance
        
    def __str__(self):
        return f"Name of the account holder : {self.name} \n Account No : {self.account_no} \n Balance : {self.balance}"
    
    
act1 = BankAccount("Prathamesh", 123456, 100000)
print(act1)
```

**Number Class – __add__()**
**Create a Number class that stores one number. Implement __add__() so that two**
**objects can be added**

```python
class Number:
    def __init__(self,num):
        self.num = num
    
    def __add__(self, other):
        return self.num + other.num

obj1 = Number(10)
obj2 = Number(20)

obj1 + obj2
```

**6. Create a Number class and implement:**
**__add__()**
**__sub__()**
**__mul__()**
**__truediv__()**
**__mod__()**

```python
class Number:
    def __init__(self,num):
        self.num = num
    
    def __add__(self, other):
        return self.num + other.num
    
    def __sub__(self, other):
        return self.num - other.num
    
    def __mul__(self, other):
        return self.num * other.num
    
    def __truediv__(self, other):
        return self.num / other.num
    
    def __mod__(self, other):
        return self.num % other.num

obj1 = Number(33)
obj2 = Number(10)

# obj1 + obj2
# obj1 - obj2
# obj1 * obj2
# obj1 / obj2
obj1 % obj2
```
