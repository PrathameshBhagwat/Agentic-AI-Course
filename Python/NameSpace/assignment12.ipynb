**What will be the output? Is x being shadowed?**

```python
x = 100
def outer():
    x = 200
    def inner():
        x = 300
        print(x)
        
    inner()
    print(x)
    
outer()
print(x)
# when the x=300 it is shadowed 
```

**What will be the output? Explain the difference between the Local and Enclosing Namespace.**

```python
def outer():
    x=10
    def inner():
        x=20
        print(x)
    
    inner()
    print(x)

outer()
# Here we can see that the inner function has the x=20  which is the local scope and then the x=10 
# which is the enclosing such that if we don't declare x and print it inside the inner it will print as it inherit the value from 
# it's parent function.
```

**What will be the output? Why does nonlocal change the value of x?**

```python
def outer():
    x=10
    def inner():
        nonlocal x 
        x=50
        print(x)
    
    inner()
    print(x)

outer()
# it changes the value of x in the outermost function it neutralizes the value of the x inside the innner function 
```
