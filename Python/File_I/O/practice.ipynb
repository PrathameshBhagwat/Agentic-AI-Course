**CSV – Student Records**

```python
import csv

total =0 
count =0
highest_mark = 0
lowest_mark = 100

highest_mark_student = ""
lowest_mark_student = ""

with open("students.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    next(reader)  # Skip the header row
    for row in reader:
        roll_no = int(row[0])
        name = row[1]
        age = int(row[2])
        mark = float(row[3])
        
        total += mark
        count += 1
        
        if mark > highest_mark:
            highest_mark = mark
            highest_mark_student = name
            
        if mark < lowest_mark:
            lowest_mark = mark
            lowest_mark_student = name
    
average_mark = total / count 

print(f"Average Mark: {average_mark:.2f}")
print(f"Highest Mark: {highest_mark} by {highest_mark_student}")
print(f"Lowest Mark: {lowest_mark} by {lowest_mark_student}")
print(f"Total Students: {count}")


```
