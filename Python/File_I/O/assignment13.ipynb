**Write a program to create a text file and store the name, age, and marks of a student in it.**

```python
name = input("What is your name? ")
age = int(input("How old are you? "))
marks = int(input("What are your marks? "))
with open("Student_details.txt", "a") as file:
    file.write(f"Name : {name}\n")
    file.write(f"Age : {age}\n")
    file.write(f"Marks : {marks}\n")
    
```

**Write a program to read the complete contents of a file and count the total number of characters.**

```python
content = "Ratan Tata is an Indian industrialist, philanthropist, and former chairman of Tata Sons. He was born on December 28, 1937, in Mumbai, India. Ratan Tata is known for his significant contributions to the Tata Group, one of India's largest conglomerates, and for his philanthropic efforts through the Tata Trusts."
with open("Ratan_Tata.txt", "a") as file:
    file.write(content)
    
with open("Ratan_Tata.txt", "r") as file:
    content = file.read()
    char_count = 0
    for char in content:
        if char!= " ":
            char_count += 1 
print(f"Number of characters: {char_count}")
```

**Write a program to read a file line by line and count the total number of lines.**

```python
with open("Student_details.txt","r") as file:
    content = file.readlines()
    print(content)
    line_count = 0
    for line in content:
        line_count += 1
print(f"Number of lines: {line_count}")
```

**Write a program to read a file and count the number of words present in it.**

```python
with open("Ratan_Tata.txt", "r") as file:
    content = file.read()
    word_count = 0
    for word in content:
        if word == " ":
            word_count += 1
print(f"Number of words: {word_count + 1}")
```

**Write a program to read a file and create another file containing only the lines that contain the word "Python".**

```python
with open("input.txt", "a") as file:
    file.write("Python is easy to learn.\n")
    file.write("Java is also popular.\n")
    file.write("I am learning Python.\n")
    file.write("C++ is a programming language.\n")
    file.write("Python is used for AI.\n")
    
with open("input.txt", "r") as file:
    content = file.readlines()
    

with open("python_lines.txt", "a") as file:
    for line in content:
        if "Python" in line:
            file.write(line)
            
with open("python_lines.txt", "r") as file:
    content = file.read()
    print(content)
```

**Write a program to read a file and find the longest word present in it.**

```python
with open("Ratan_Tata.txt", "r") as file:
    content = file.read()
    longest_word = ""
    for word in content.split():
        if len(word) > len(longest_word):
            longest_word = word
print(f"The longest word : {longest_word}")
```

**Write a program to read a file containing numbers and calculate their sum, average, maximum, and minimum.**

```python
with open("Numbers.txt", "a") as file:
    for i in range(1,20):
        file.write(f"{i}\n")
    
with open("Numbers.txt", "r") as file:
    content = []
    for line in file:
        content.append(int(line.strip()))
    total = 0
    average = 0
    maximum = 0
    minimum = 0 
    for number in content:
        total = total + number
        
    average = total / len(content)
    maximum = max(content)
    minimum = min(content)
print(f"Sum: {total}")
print(f"Average: {average}")
print(f"Maximum: {maximum}")
print(f"Minimum: {minimum}")
        
```
