**Write a program to count the number of vowels and consonants in a given string.**

```python
name = input("What is your name? ")
consonants = 0
vowels = 0

for i in name:
    if i.lower() in "aeiou":
        vowels += 1
    elif i.isalpha():
        consonants += 1

print("consonants : ", consonants ,"vowels : ", vowels)
```

**Write a program to check whether a given string is a Palindrome**

```python
word = input("Enter a word: ")
word = word.strip().lower()
reverse_word = word[::-1]
if word == reverse_word:
    print("The word is a palindrome.")
else:
    print("The word is not a palindrome.")
```

**Write a program to reverse a string without using slicing ([::-1]) or built-in functions.**

```python
word = input("Enter a word: ")
reverse_word = " "
for i in word:
    reverse_word = i + reverse_word
print("The reverse of the word is:", reverse_word)
```

**Write a program to count the occurrence of each character in a string.**

```python
word = input("Enter a word: ")
word = word.strip().lower()
for i in word:
    if word.count(i)>0:
        print(i, ":", word.count(i))
```

**Write a program to remove all duplicate characters from a string while maintaining their original order.**

```python
word = input("Enter a word: ")
word = word.strip().lower()
without_duplicates = ""
for i in word:
    if i not in without_duplicates:
        without_duplicates = without_duplicates + i
print("The word without duplicates is:", without_duplicates)
```

**Write a program to find the longest word in a sentence**

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
longest_word = ""
for word in words:
    if len(word) > len(longest_word):
        longest_word = word
print("The longest word is:", longest_word)
```

**Write a program to capitalize the first letter of every word without using the title()**

```python
sentence = input("Enter a sentence: ")

words = sentence.split()
new_sentence = []

for word in words:
    if word and words:
        new_sentence.append(word.capitalize())

print(" ".join(new_sentence))
```

**Write a program to count the total number of words in a sentence without using the split() method.**

```python
sentence = input("Enter a sentence: ")
words = 0
for i in sentence:
    if i == " ":
        words += 1
print(f"Number of words in the sentence: {words}")
```
